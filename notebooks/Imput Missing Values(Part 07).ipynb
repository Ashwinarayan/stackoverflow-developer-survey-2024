### Import Required Libraries


```python
import pandas as pd
```

### Load the Dataset Into a Dataframe


```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(file_path)

df.head()
```

### Step 3. Finding and Removing Duplicates
##### Identify duplicate rows in the dataset.


```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

##### Remove the duplicate rows from the dataframe.



```python
df_unique = df.drop_duplicates(keep='first')
# Check how many rows were removed
removed_count = len(df) - len(df_unique)
print("Number of duplicates removed:", removed_count)
```

### Step 4: Finding Missing Values
##### Find the missing values for all columns.


```python
# Check missing values by column
missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values)
```

Find out how many rows are missing in the column RemoteWork.


```python
# Check missing values by column
missing_values = df['RemoteWork'].isnull().sum()
print("Missing values per column:\n", missing_values)
```

### Step 5. Imputing Missing Values
##### Find the value counts for the column RemoteWork.


```python
# Count how many times each RemoteWork category appears
remote_work_counts = df['RemoteWork'].value_counts()

print("Value counts for RemoteWork column:")
print(remote_work_counts)
```

##### Identify the most frequent (majority) value in the RemoteWork column.



```python
# Find which RemoteWork category appears most often
most_common_remote = df['RemoteWork'].mode()[0]  # Most frequent value
print("Most frequent RemoteWork value:", most_common_remote)
```

##### Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
df.fillna({'RemoteWork': most_common_remote}, inplace=True)

# Verify
print("Missing after imputation:", df['RemoteWork'].isnull().sum())
```

##### Check for any compensation-related columns and describe their distribution.



```python
# Summary statistics of compensation
comp_stats = df['ConvertedCompYearly'].describe()
print("Summary statistics for ConvertedCompYearly:\n", comp_stats)
```
