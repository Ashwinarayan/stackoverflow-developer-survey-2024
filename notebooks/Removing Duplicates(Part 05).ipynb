### Load the Dataset into a DataFrame



```python
# Define the URL of the dataset
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/VYPrOu0Vs3I0hKLLjiPGrA/survey-data-with-duplicate.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Identifying Duplicate Rows


**Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
duplicate_rows = df[df.duplicated()]
duplicate_rows.head()
```

### Removing Duplicate Rows


**Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
df_unique = df.drop_duplicates(keep='first')

# Check how many rows were removed
removed_count = len(df) - len(df_unique)
print("Number of duplicates removed:", removed_count)
```

### Handling Missing Values


**Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
# Check missing values by column
missing_values = df.isnull().sum()
print("Missing values per column:\n", missing_values)
```

```python
# Count missing values again to confirm
print("Missing before imputation:\n", df['EdLevel'].isnull().sum())

# Impute missing values in 'EdLevel' with the most common value
most_freq = df['EdLevel'].mode()[0]       # .mode() gives all most frequest values list or single value. [0] gives 1st value from list even if there are 10, 20, 30 vales are most frequent it gives 10 coz we use [0]
df.fillna({'EdLevel': most_freq}, inplace=True)

# Check again after imputation
print("Missing after imputation:\n", df['EdLevel'].isnull().sum())
```

### Normalizing Compensation Data


**Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
# Check missing values in ConvertedCompYearly
missing_comp = df['ConvertedCompYearly'].isnull().sum()
print("Missing values in ConvertedCompYearly:", missing_comp)
```

2. Decide How to Handle Missing Values

Since ConvertedCompYearly is a numerical compensation column, there are a few common strategies to handle missing values:

🧠 Options for Numeric Columns

Drop rows if only a small percentage is missing and you don’t want to impute. 

Impute with a statistic like mean or median:

Mean: average compensation — useful if the distribution has no extreme outliers.

Median: middle compensation — better when data is skewed or has outliers. 

For compensation (a numeric financial value), the median is often preferable because salary distributions tend to be skewed by high earners. But you can choose based on your analysis.

```python
# Calculate median of ConvertedCompYearly
median_comp = df['ConvertedCompYearly'].median()

# Fill missing values with median
#df['ConvertedCompYearly'].fillna(median_comp, inplace=True)
df.fillna({'ConvertedCompYearly': median_comp}, inplace=True)
# Verify
print("Missing after imputation:", df['ConvertedCompYearly'].isnull().sum())
```
